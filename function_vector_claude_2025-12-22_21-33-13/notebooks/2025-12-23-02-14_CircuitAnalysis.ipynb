```python
# First, let's explore the repository structure
import os

repo_root = '/net/scratch2/smallyan/leela-logit-lens_eval'

def list_directory_structure(path, prefix="", max_depth=3, current_depth=0):
    if current_depth >= max_depth:
        return
    
    try:
        items = sorted(os.listdir(path))
        for item in items:
            item_path = os.path.join(path, item)
            if os.path.isdir(item_path):
                print(f"{prefix}📁 {item}/")
                list_directory_structure(item_path, prefix + "  ", max_depth, current_depth + 1)
            else:
                print(f"{prefix}📄 {item}")
    except PermissionError:
        print(f"{prefix}[Permission Denied]")

print(f"Repository structure of: {repo_root}\n")
list_directory_structure(repo_root)
```

```python
# Now I understand the research. Let me summarize the key findings:
# 1. This is a "logit lens" analysis of Leela Chess Zero (Lc0) neural network
# 2. The main model used: lc0-original.onnx (T82-768x15x24h transformer with 15 layers)
# 3. Key finding: Three-phase progression in the neural network:
#    - Early layers (0-5): rapid capability gains
#    - Middle layers (6-10): plateau
#    - Late layers (11-14): sharp strengthening
# 4. Method: Zero ablation - projecting intermediate representations through the policy head

# For generalization testing, I need to:
# GT1: Test on a NEW MODEL (not lc0-original.onnx or lc0.onnx that are used in the paper)
# GT2: Test on NEW DATA (not from the original puzzle dataset)
# GT3: Check if the logit lens method can be applied to another similar task

# Let's check if there's an evaluation folder already
eval_path = os.path.join(repo_root, 'evaluation')
if not os.path.exists(eval_path):
    os.makedirs(eval_path)
    print(f"Created evaluation directory: {eval_path}")
else:
    print(f"Evaluation directory already exists: {eval_path}")
    for item in os.listdir(eval_path):
        print(f"  {item}")
```
