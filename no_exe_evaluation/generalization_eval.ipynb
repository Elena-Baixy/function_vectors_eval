# Generalization Evaluation (Read-Only Static Inspection)

This notebook evaluates the generalizability of the Function Vectors research based on static inspection of the repository, without executing any code.

**Repository:** `/net/scratch2/smallyan/function_vectors_eval`

**Evaluation Criteria:**
- GT1: Generalization to a New Model
- GT2: Generalization to New Data  
- GT3: Method/Specificity Generalizability

## GT1: Generalization to a New Model

### Evidence Found

**Empirical multi-model testing is present:**

From `plan.md`:
- "Test function vectors across models (GPT-J 6B to Llama 2 70B)"
- "Causal mediation analysis across models: Model architecture and size (GPT-J 6B, GPT-NeoX 20B, Llama 2 7B/13B/70B)"
- "Top 10-100 attention heads (scaled by model size) with highest AIE cluster in middle layers across all models"

From `src/utils/extract_utils.py` (lines 387-452):
- Pre-computed universal head sets are provided for: GPT-J, Llama-2-7B, Llama-2-13B, Llama-2-70B, GPT-NeoX
- The function `compute_universal_function_vector()` contains model-specific head indices

From `src/utils/model_utils.py`:
- `load_gpt_model_and_tokenizer()` supports multiple architectures: GPT-2-XL, GPT-J, GPT-NeoX, Pythia, Gemma, Llama, OLMo

### Missing Justification

**No explicit mechanistic reasoning is provided** explaining WHY function vectors should transfer to different models. The repository:

1. Does not explain what architectural properties enable this transfer
2. Does not identify invariances that would predict generalization
3. Does not provide theoretical grounding for cross-model applicability
4. Does not specify conditions under which model transfer would fail

The observation that "top heads cluster in middle layers across all models" is an empirical finding, not a mechanistic explanation for generalization.

### Verdict: **FAIL**

While the repository demonstrates empirical testing across multiple models, it does not explicitly explain WHY the findings should transfer based on architectural properties, invariances, or mechanistic reasoning.

## GT3: Method/Specificity Generalizability

### Method Identification

The repository proposes a **new method** for extracting function vectors:

From `plan.md`:
- "Apply causal mediation analysis to identify attention heads with highest average indirect effect (AIE)"
- "Extract function vectors by summing task-conditioned mean outputs of top causal attention heads"

Core method components from code:
1. `get_mean_head_activations()` - computes average attention head activations across ICL prompts
2. `compute_indirect_effect()` - performs causal mediation analysis to find influential heads
3. `compute_function_vector()` - combines top heads into a function vector
4. `function_vector_intervention()` - applies the FV via residual stream addition

### Evidence Found

The method is applied to:
- Multiple models (GPT-J, Llama variants, GPT-NeoX)
- Multiple task types (abstractive, extractive, generative)
- Multiple contexts (ICL, zero-shot, natural text)

### Missing Justification

**No explicit argument about method generalizability:**

1. No specification of which properties are required vs. incidental
2. No discussion of what task types the method should/should not work for
3. No theoretical basis for expecting the method to work on new tasks
4. No boundary conditions or failure modes described

The documentation and code show the method being applied broadly but do not explain why it would generalize to other settings.

### Verdict: **FAIL**

A new method is proposed and applied empirically, but there is no explicit justification for why it would generalize to other tasks or settings.

## Summary

### Overall Generalizability Assessment

Based on read-only static inspection, the Function Vectors repository demonstrates **extensive empirical testing** across:
- Multiple models (5+ architectures of varying sizes)
- Multiple tasks (40+ diverse ICL tasks)
- Multiple contexts (ICL, zero-shot, shuffled-label, natural text)

However, the repository **fails all three generalization criteria** because it lacks:

1. **Mechanistic reasoning**: No explanation of WHY function vectors work or should transfer across models/tasks
2. **Scope conditions**: No explicit assumptions about when findings would/would not hold
3. **Theoretical grounding**: No principled basis for expecting generalization beyond tested settings

The approach follows an empirical paradigm where generalization is demonstrated through testing rather than explained through theory. While the breadth of experiments provides confidence in the findings, the absence of explicit justification means generalization cannot be assumed for untested models, data, or tasks without additional verification.

**Key limitation**: Assertions that function vectors "work across models" or "generalize across tasks" are supported by experimental results but not by mechanistic or theoretical arguments visible in the repository documentation.
