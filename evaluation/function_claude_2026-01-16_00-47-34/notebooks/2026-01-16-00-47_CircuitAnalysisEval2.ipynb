## Project Goal (from plan.md)
Investigate whether autoregressive transformer language models develop compact vector representations of input-output functions (function vectors) within their hidden states during in-context learning.

## Key Methodology
1. Apply causal mediation analysis to identify attention heads with highest average indirect effect (AIE)
2. Extract function vectors by summing task-conditioned mean outputs of top causal attention heads
3. Test function vectors across models and diverse ICL tasks
