## Key Findings from Documentation (ICLR 2024 Paper)

Based on reading the documentation PDF:

### Main Claims:
1. **Function Vectors (FVs)** are compact vector representations of input-output tasks found in transformer hidden states during ICL
2. FVs are extracted from a small number of attention heads (10-100 depending on model size) identified through causal mediation analysis
3. FVs work best when added at early-middle layers (approximately L/3)
4. FVs are portable across contexts: shuffled-label ICL, zero-shot, different templates, natural text

### Key Results Reported:
- **Shuffled-label context**: GPT-J+FV achieves 90.8% vs 39.1% baseline
- **Zero-shot context**: GPT-J+FV achieves 57.5% vs 5.5% baseline
- FVs robust across 20 different templates
- Natural text portability: Antonym FV achieves 55-68% vs 0-3% baseline
- Vocabulary reconstruction underperforms original FVs
- Vector algebra composition works for some tasks

```python
# Read the complete notebook content and outputs
def print_cell_with_outputs(notebook, cell_idx):
    cell = notebook['cells'][cell_idx]
    print(f"\n{'='*60}")
    print(f"CELL {cell_idx} ({cell['cell_type']})")
    print('='*60)
    print(''.join(cell['source']))
    if 'outputs' in cell:
        print("\n--- OUTPUTS ---")
        for output in cell['outputs']:
            if output['output_type'] == 'stream':
                print(''.join(output.get('text', [])))
            elif output['output_type'] == 'execute_result':
                data = output.get('data', {})
                if 'text/plain' in data:
                    print(''.join(data['text/plain']))
            elif output['output_type'] == 'display_data':
                data = output.get('data', {})
                if 'text/plain' in data:
                    print(''.join(data['text/plain']))

# Print all code cells with their outputs
for i in range(len(notebook['cells'])):
    print_cell_with_outputs(notebook, i)
```

```python
# The documentation is a published ICLR 2024 paper with specific quantitative results
# The notebook (fv_demo.ipynb) has no recorded outputs - it's a demonstration template
# Let's verify by checking the code implementation matches the methodology described

# Check the key implementation details against the plan and documentation

# From Plan:
plan_claims = {
    "FV_extraction_method": "Apply causal mediation analysis to identify attention heads with highest AIE",
    "FV_definition": "Sum task-conditioned mean outputs of top causal attention heads",
    "evaluation_contexts": ["shuffled-label ICL", "zero-shot", "different ICL templates", "natural text"],
    "evaluation_metric": "Top-1 accuracy",
    "layer_for_intervention": "approximately L/3",
    "results_shuffled_label": "GPT-J+FV achieves 90.8% vs 39.1% baseline",
    "results_zero_shot": "GPT-J+FV achieves 57.5% vs 5.5% baseline"
}

# From Documentation PDF - Key Equations/Methods:
doc_methodology = """
1. Equation 2: Mean task-conditioned activations ā^t_{ℓj} = (1/|P_t|) Σ a_{ℓj}(p^t_i)
2. Equation 3: CIE(a_{ℓj} | p̃^t_i) = f(p̃^t_i | a_{ℓj} := ā^t_{ℓj})[y_iq] - f(p̃^t_i)[y_iq]
3. Equation 4: AIE(a_{ℓj}) = averaged CIE across all tasks and prompts
4. Equation 5: v_t = Σ_{a_{ℓj} ∈ A} ā^t_{ℓj} (sum of top attention head outputs)
"""

print("Plan claims:")
for key, value in plan_claims.items():
    print(f"  {key}: {value}")

print("\nDocumentation methodology:")
print(doc_methodology)
```
