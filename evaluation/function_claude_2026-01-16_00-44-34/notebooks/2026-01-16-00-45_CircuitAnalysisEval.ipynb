### Evaluation Approach:
We will evaluate each code block in the fv_demo.ipynb notebook by executing them in order.

```python
# Set up path to the repo
repo_path = '/net/scratch2/smallyan/function_vectors_eval'
os.chdir(repo_path)
sys.path.append(repo_path)

# Initialize evaluation tracking
evaluation_results = []

def add_eval_result(file_name, block_id, runnable, correct_impl, redundant, irrelevant, error_note=""):
    """Add an evaluation result to the tracking list"""
    evaluation_results.append({
        'file': file_name,
        'block_id': block_id,
        'runnable': runnable,
        'correct_implementation': correct_impl,
        'redundant': redundant,
        'irrelevant': irrelevant,
        'error_note': error_note
    })
    print(f"Evaluated: {file_name} - {block_id}: R={runnable}, C={correct_impl}, Red={redundant}, Irr={irrelevant}")
```
