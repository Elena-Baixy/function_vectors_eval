```python
# Let's also check the demo notebook for reference results
demo_notebook_path = os.path.join(ORIGINAL_REPO, 'notebooks', 'fv_demo.ipynb')
if os.path.exists(demo_notebook_path):
    import json
    with open(demo_notebook_path, 'r') as f:
        demo_notebook = json.load(f)
    
    print("=== Demo Notebook Contents ===")
    print(f"Number of cells: {len(demo_notebook['cells'])}")
    
    # Extract markdown and code cells
    for i, cell in enumerate(demo_notebook['cells']):
        if cell['cell_type'] == 'markdown':
            content = ''.join(cell['source'])
            if len(content) > 0:
                print(f"\n--- Cell {i} (Markdown) ---")
                print(content[:500] + "..." if len(content) > 500 else content)
        elif cell['cell_type'] == 'code':
            # Check for outputs
            if cell.get('outputs'):
                for output in cell['outputs']:
                    if output.get('output_type') == 'stream':
                        text = ''.join(output.get('text', []))
                        if text.strip():
                            print(f"\n--- Cell {i} (Code Output) ---")
                            print(text[:1000] + "..." if len(text) > 1000 else text)
                    elif output.get('output_type') == 'execute_result':
                        data = output.get('data', {})
                        if 'text/plain' in data:
                            text = ''.join(data['text/plain'])
                            print(f"\n--- Cell {i} (Execute Result) ---")
                            print(text[:500] + "..." if len(text) > 500 else text)
else:
    print("Demo notebook not found")
```
